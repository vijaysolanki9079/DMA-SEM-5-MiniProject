```python
import pandas as pd
import numpy as np

df = pd.read_csv('/content/Acad-Data.csv', encoding='latin-1')
display(df.head())

df.info()
df.describe()
df.isnull().sum()
```

```python
# Removing bad words
import pandas as pd

bad_words = ['backshot', "haven't given exams", 'hale', "i'm", 'papers', 'fresher']
df = df[~df.apply(lambda row: row.astype(str)
                               .str.lower()
                               .apply(lambda x: any(bad_word.lower() in x for bad_word in bad_words))
                               .any(), axis=1)]

df.reset_index(drop=True, inplace=True)

df.head()
```

```python
# Defining a dictionary
new_column_names = {
    'Name :\n\nNote: If you donât want to share your real name, just use a fun/unique name like Jacky53, Munja89, etc. ð Please donât leave it blank with random stuff or bad words â keep it decent oneð': 'Name',
    'Year of Study ': 'Year',
    'Branch \n\nExample: CSE, ME, CE...etc': 'Branch',
    'What type of area was your school located in? ': 'School_Area',
    'What was your percentage/cgpa in Class 10th?': '10th',
    'What was your percentage/cgpa in Class 12th?': '12th',
    'Were you involved in any activities during school (Class 9th12th)?': 'School_Activities',
    'How did you choose your stream after Class 10th?': 'Stream_Choice',
    'What is your current CGPA?': 'Current_CGPA',
    'What was your CGPA in previous Year/Semester?\n\n1st yr: As per Semester \nOther yrs: As per Year ': 'Previous_CGPA',
    'How has your CGPA changed over time?': 'CGPA_Change',
    'What are the main reasons affecting your CGPA? (You can select multiple)': 'CGPA_Reasons',
    'Rate our colleges involvement in your personal & academic growth? ': 'College_Support',
    'Are you pursuing any side interests or passions in college?  ': 'Side_Interests',
    'If above you selected "Yes". Please mention.': 'Side_Interests_Details',
    'What is your main motivation for studying in college?': 'College_Motivation',
    'Are you satisfied with your academic journey so far?': 'Academic_Satisfaction',
    'Gender ': 'Gender'
}

df.columns.values[0] = 'Name'
df.columns.values[6] = 'School_Activities'
df.columns.values[12] = 'Rate our College'

df = df.rename(columns=new_column_names)
display(df.head())
```

```python
df = df[~df.apply(lambda row: row.astype(str).str.contains('backshot', case=False).any(), axis=1)]
df.reset_index(drop=True, inplace=True)
df.head()
```

```python
df['Year'] = df['Year'].astype(str).str.extract(r'(\d+)').astype(float).astype('Int64')
display(df.head())
```

```python
df['School_Area'] = df['School_Area'].str.replace(r'[^a-zA-Z\s-]', '', regex=True).str.strip()
display(df.head())
```

```python
df = df.dropna(how='all')
display(df.head())
```

```python
df['Branch'] = df['Branch'].astype(str).str.strip().str.upper().replace('DSE', 'DS')
unique_branch_values = df['Branch'].unique()
print("Unique values in the 'Branch' column:")
for branch in unique_branch_values:
    print(branch)
```

```python
df['Branch'] = df['Branch'].astype(str).str.strip().str.upper().replace('DSE', 'DS')

valid_branches = ['CSE', 'ECE', 'IT', 'ME', 'EE', 'CE', 'CH', 'BT', 'IPE', 'ICE', 'DS', 'MNC', 'MCE']

manual_corrections = {
    'YOUR FATHERS BRANCH': 'CSE',
    'MCE OR MNC': 'MNC',
    '': None,
}

df['Branch'] = df['Branch'].replace(manual_corrections)

df.loc[~df['Branch'].isin(valid_branches), 'Branch'] = 'UNKNOWN'

unique_branch_values = df['Branch'].unique()
print("✅ Cleaned unique values in the 'Branch' column:")
for branch in unique_branch_values:
    print(branch)
```

**Look at Columns**

Categorize the '10th' and '12th' columns into grades based on percentage ranges and visualize the distribution of these grades.

```python
percentage_ranges = {
    'Above 90%': '>90%',
    'Between 80% - 90%': '80-90%',
    'Between 60% - 80%': '60-80%',
    'Under 60%': '<60%'
}

print("Percentage/CGPA ranges for categorization:")
for key, value in percentage_ranges.items():
    print(f"- {key}: {value}")
```

```python
import re
import unidecode

def clean_activity(text):
    text = str(text).lower()
    text = unidecode.unidecode(text)
    text = re.sub(r'[^\w\s,]', '', text)
    text = re.sub(r'\s+', ' ', text).strip()
    return text

df['School_Activities'] = df['School_Activities'].apply(clean_activity)

options = {
    'Sports': ['sports'],
    'Arts': ['arts', 'arts music/dance/drama etc'],
    'Tech': ['coding / tech competitions', 'tech'],
    'Academics': ['olympiads / quiz / other academics', 'academics'],
    'None': ['none']
}

for col, keywords in options.items():
    df[col] = df['School_Activities'].apply(lambda x: 1 if any(k in x for k in keywords) else 0)

df.drop(columns=['School_Activities'], inplace=True)

df.head(20)
```

**Stream_Choice**

Analyze and visualize the distribution of the stream_choice column to understand the reasons for choosing a stream after Class 10th.

```python
# Clean column names (remove spaces/newlines)
df.columns = df.columns.str.strip().str.replace('\n', '_')

def clean_stream(text):
    text = str(text).lower()
    text = unidecode.unidecode(text)
    text = re.sub(r'[^\w\s,()/]', '', text)
    text = re.sub(r'\s+', ' ', text).strip()
    return text

df['Stream_Choice'] = df['Stream_Choice'].apply(clean_stream)

stream_options = {
    'Own_Interest': ['own interest', 'interest', 'own'],
    'Parents_Suggestion': ['parents', 'relatives', 'sharma ji'],
    'Peer_Pressure': ['peer pressure', 'peer'],
    'No_Choice': ['no choice', 'fixed']
}

for col, keywords in stream_options.items():
    df[col] = df['Stream_Choice'].apply(lambda x: 1 if any(k in x for k in keywords) else 0)

df.drop(columns=['Stream_Choice'], inplace=True)
df.head(20)
```

```python
# Clean column names
df.columns = df.columns.str.strip().str.replace('\n', '_')

def clean_reason(text):
    text = str(text).lower()
    text = unidecode.unidecode(text)
    text = re.sub(r'[^\w\s,()]', '', text)
    text = re.sub(r'\s+', ' ', text).strip()
    return text

df['CGPA_Reasons'] = df['CGPA_Reasons'].apply(clean_reason)

cgpa_reasons_options = {
    'Interest_in_studies': ['interest in studies', 'interest'],
    'Time_management': ['time management', 'time management issues'],
    'Attendance': ['attendance rules', 'attendance'],
    'Peer_influence': ['friends', 'peer influence', 'peer'],
    'Health_issues': ['health issues', 'health'],
    'Lack_of_interest_in_branch': ['lack of interest in branch', 'lack of interest'],
    'Other_passions': ['focus on other passions', 'projects', 'other passions']
}

for col, keywords in cgpa_reasons_options.items():
    df[col] = df['CGPA_Reasons'].apply(lambda x: 1 if any(k in x for k in keywords) else 0)

df.drop(columns=['CGPA_Reasons'], inplace=True)

df.head(20)
```

**CGPA Change**

```python
import pandas as pd
import re
import unidecode

def clean_change(text):
    text = str(text).lower()
    text = unidecode.unidecode(text)
    text = re.sub(r'[^a-z\s]', '', text)
    text = text.strip()

    if 'mix' in text:
        return 'Mix'
    elif 'improved' in text:
        return 'Improved'
    elif 'declined' in text:
        return 'Declined'
    else:
        return 'Other'

df['CGPA_Change'] = df['CGPA_Change'].apply(clean_change)
df.head(10)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")

# Pie Chart
plt.figure(figsize=(7,7))
df['CGPA_Change'].value_counts().plot.pie(autopct='%1.1f%%', colors=sns.color_palette('Set2'), startangle=90)
plt.title('Distribution of CGPA Change')
plt.ylabel('')
plt.show()

# Bar Chart
plt.figure(figsize=(8,6))
sns.countplot(x='CGPA_Change', data=df, palette='Set2', order=['Improved','Declined','Mix','Other'])
plt.title('CGPA Change Distribution')
plt.ylabel('Number of Students')
plt.xlabel('CGPA Change')
plt.show()
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.preprocessing import OneHotEncoder
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns

# Define target and feature columns
target_column = 'Academic_Satisfaction'

# Exclude identifier, free text, and duplicate-meaning columns
exclude_columns = ['Name', 'Side_Interests_Details', '10th', '12th']

# Use all other relevant columns as features
feature_columns = [col for col in df.columns if col not in exclude_columns + [target_column]]

X = df[feature_columns]
y = df[target_column]

# Identify categorical and numerical features
categorical_features = X.select_dtypes(include='object').columns.tolist()
numerical_features = X.select_dtypes(include=['int64', 'float64', 'Int64']).columns.tolist()

print(f"Target variable: {target_column}")
print(f"Selected features: {feature_columns}")
print(f"Categorical features to encode: {categorical_features}")
print(f"Numerical features: {numerical_features}")
```

### One-Hot Encode Categorical Features

We need to convert categorical features into a format that machine learning algorithms can understand. One-hot encoding creates new binary columns for each category.

```python
# Apply One-Hot Encoding to categorical features
encoder = OneHotEncoder(handle_unknown='ignore', sparse_output=False)
X_encoded = encoder.fit_transform(X[categorical_features])

# Create a DataFrame from the encoded features
X_encoded_df = pd.DataFrame(X_encoded, columns=encoder.get_feature_names_out(categorical_features))

# Combine numerical features with the one-hot encoded categorical features
X_final = pd.concat([X[numerical_features].reset_index(drop=True), X_encoded_df], axis=1)

# --- FIX START: Clean column names for XGBoost compatibility ---
X_final.columns = [
    re.sub(r'[<>\[\]]', '_', col)
    for col in X_final.columns
]
X_final.columns = [
    re.sub(r'[^a-zA-Z0-9_]', '', col) # Remove any remaining non-alphanumeric chars (except underscore)
    for col in X_final.columns
]
# --- FIX END ---

# Check the first few rows of the processed features
display(X_final.head())
```

### Split Data into Training and Testing Sets

Now, we'll split our data into training and testing sets to evaluate the model's performance on unseen data. A common split is 80% for training and 20% for testing.

```python
# Split data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X_final, y, test_size=0.2, random_state=42)

print(f"X_train shape: {X_train.shape}")
print(f"X_test shape: {X_test.shape}")
print(f"y_train shape: {y_train.shape}")
print(f"y_test shape: {y_test.shape}")
```

### Train the Decision Tree Model

We'll use a `DecisionTreeClassifier` to train our model. We can adjust parameters like `max_depth` to control the complexity of the tree and prevent overfitting.

```python
# Initialize and train the Decision Tree Classifier
dtree_model = DecisionTreeClassifier(random_state=42, max_depth=5) # Limiting max_depth for better interpretability
dtree_model.fit(X_train, y_train)

print("Decision Tree Model trained successfully!")
```

### Evaluate the Model

```python
import pandas as pd
import numpy as np
import re
import unidecode
import matplotlib.pyplot as plt
import seaborn as sns

# Re-load and initial cleaning based on previous cells
try:
    df = pd.read_csv('/content/Acad-Data.csv', encoding='latin-1')
except FileNotFoundError:
    print("Error: The file '/content/Acad-Data.csv' was not found. Please ensure the CSV file is uploaded to the /content/ directory in your Colab environment to proceed with the analysis.")
    df = pd.DataFrame() # Initialize an empty DataFrame to prevent immediate crashes in subsequent steps

# Only proceed with preprocessing if the DataFrame was successfully loaded (i.e., not empty)
if not df.empty:
    # Apply initial filtering and reset_index (from cell E7Nodo282hXY)
    bad_words = ['backshot', "haven't given exams", 'hale', "i'm", 'papers', 'fresher']
    df = df[~df.apply(lambda row: row.astype(str)
                                   .str.lower()
                                   .apply(lambda x: any(bad_word.lower() in x for bad_word in bad_words))
                                   .any(), axis=1)]
    df.reset_index(drop=True, inplace=True)

    # Renaming columns (from cell KV8oDupBmcyH and subsequent cleaning cells)
    new_column_names = {
        'Name :\n\nNote: If you donât want to share your real name, just use a fun/unique name like Jacky53, Munja89, etc. ð\u009f\u0091\u0089 Please donât leave it blank with random stuff or bad words â keep it decent oneð\u009f\u0099\u0082': 'Name',
        'Year of Study ': 'Year',
        'Branch \n\nExample: CSE, ME, CE...etc': 'Branch',
        'What type of area was your school located in? ': 'School_Area',
        'What was your percentage/cgpa in Class 10th?': '10th',
        'What was your percentage/cgpa in Class 12th?': '12th',
        'Were you involved in any activities during school (Class 9thâ12th)?': 'School_Activities',
        'How did you choose your stream after Class 10th?': 'Stream_Choice',
        'What is your current CGPA?': 'Current_CGPA',
        'What was your CGPA in previous Year/Semester?\n\n1st yr: As per Semester \nOther yrs: As per Year ': 'Previous_CGPA',
        'How has your CGPA changed over time?': 'CGPA_Change',
        'What are the main reasons affecting your CGPA? (You can select multiple)': 'CGPA_Reasons',
        'Rate our collegeâs involvement in your personal & academic growth? ': 'College_Support',
        'Are you pursuing any side interests or passions in college?  ': 'Side_Interests',
        'If above you selected "Yes". Please mention.': 'Side_Interests_Details',
        'What is your main motivation for studying in college?': 'College_Motivation',
        'Are you satisfied with your academic journey so far?': 'Academic_Satisfaction',
        'Gender ': 'Gender'
    }
    df = df.rename(columns=new_column_names)

    # Handle explicit renames that happened later to clean column names
    df.columns = df.columns.str.strip().str.replace('\n', '_') # From cell iGGpChfovraL
    if 'Rate our College' in df.columns: # From cell hX94FyAmAIea
        df.rename(columns={'Rate our College': 'Rate_College'}, inplace=True)
    # Ensure 'College_Support' also gets renamed to 'Rate_College' if it still exists from initial rename
    if 'College_Support' in df.columns:
         df.rename(columns={'College_Support': 'Rate_College'}, inplace=True)

    # Cleaning Year (from cell K66BjVG0m7fH)
    df['Year'] = df['Year'].astype(str).str.extract(r'(\d+)').astype(float).astype('Int64')

    # Cleaning School_Area (from cell _zdWDboimcqp)
    df['School_Area'] = df['School_Area'].str.replace(r'[^a-zA-Z\s-]', '', regex=True).str.strip()

    # Dropping all-NaN rows (from cell vp2n_gZFnnmY)
    df = df.dropna(how='all')

    # Cleaning Branch (from cell eDaqJhMMnqxw)
    df['Branch'] = df['Branch'].astype(str).str.strip().str.upper().replace('DSE', 'DS')
    valid_branches = ['CSE', 'ECE', 'IT', 'ME', 'EE', 'CE', 'CH', 'BT', 'IPE', 'ICE', 'DS', 'MNC', 'MCE']
    manual_corrections = {
        'YOUR FATHERS BRANCH': 'CSE',
        'MCE OR MNC': 'MNC',
        '': None,
    }
    df['Branch'] = df['Branch'].replace(manual_corrections)
    df.loc[~df['Branch'].isin(valid_branches), 'Branch'] = 'UNKNOWN'

    # Create 10th and 12th grade columns (from cell ePiHVJ8Hobu9)
    percentage_ranges = {
        'Above 90%': '>90%',
        'Between 80% - 90%': '80-90%',
        'Between 60% - 80%': '60-80%',
        'Under 60%': '<60%'
    }
    df['10th_grade'] = df['10th'].map(percentage_ranges).fillna('Unknown')
    df['12th_grade'] = df['12th'].map(percentage_ranges).fillna('Unknown')

    # Clean School_Activities and create binary flags (from cell wN915Tvoo37P)
    def clean_activity(text):
        text = str(text).lower()
        text = unidecode.unidecode(text)
        text = re.sub(r'[^\w\s,]', '', text) # Corrected regex based on previous code
        text = re.sub(r'\s+', ' ', text).strip()
        return text
    df['School_Activities'] = df['School_Activities'].apply(clean_activity)
    activity_options = {
        'Sports': ['sports'],
        'Arts': ['arts', 'arts music/dance/drama etc'],
        'Tech': ['coding / tech competitions', 'tech'],
        'Academics': ['olympiads / quiz / other academics', 'academics'],
        'None': ['none']
    }
    for col, keywords in activity_options.items():
        df[col] = df['School_Activities'].apply(lambda x: 1 if any(k in x for k in keywords) else 0)
    df.drop(columns=['School_Activities'], inplace=True)


    # Clean Stream_Choice and create binary flags (from cell iGGpChfovraL)
    def clean_stream(text):
        text = str(text).lower()
        text = unidecode.unidecode(text)
        text = re.sub(r'[^\w\s,()/]', '', text) # Corrected regex based on previous code
        text = re.sub(r'\s+', ' ', text).strip()
        return text
    df['Stream_Choice'] = df['Stream_Choice'].apply(clean_stream)
    stream_options = {
        'Own_Interest': ['own interest', 'interest', 'own'],
        'Parents_Suggestion': ['parents', 'relatives', 'sharma ji'],
        'Peer_Pressure': ['peer pressure', 'peer'],
        'No_Choice': ['no choice', 'fixed']
    }
    for col, keywords in stream_options.items():
        df[col] = df['Stream_Choice'].apply(lambda x: 1 if any(k in x for k in keywords) else 0)
    df.drop(columns=['Stream_Choice'], inplace=True)


    # Clean CGPA_Reasons and create binary flags (from cell QoI-3oss0nrz)
    def clean_reason(text):
        text = str(text).lower()
        text = unidecode.unidecode(text)
        text = re.sub(r'[^\w\s,()]', '', text) # Corrected regex based on previous code
        text = re.sub(r'\s+', ' ', text).strip()
        return text
    df['CGPA_Reasons'] = df['CGPA_Reasons'].apply(clean_reason)
    cgpa_reasons_options = {
        'Interest_in_studies': ['interest in studies', 'interest'],
        'Time_management': ['time management', 'time management issues'],
        'Attendance': ['attendance rules', 'attendance'],
        'Peer_influence': ['friends', 'peer influence', 'peer'],
        'Health_issues': ['health issues', 'health'],
        'Lack_of_interest_in_branch': ['lack of interest in branch', 'lack of interest'],
        'Other_passions': ['focus on other passions', 'projects', 'other passions']
    }
    for col, keywords in cgpa_reasons_options.items():
        df[col] = df['CGPA_Reasons'].apply(lambda x: 1 if any(k in x for k in keywords) else 0)
    df.drop(columns=['CGPA_Reasons'], inplace=True)


    # Convert Current_CGPA to numeric and create CGPA_Range (from cell UKtAmByD5C38)
    df['Current_CGPA'] = pd.to_numeric(df['Current_CGPA'], errors='coerce')
    df = df.dropna(subset=['Current_CGPA']).reset_index(drop=True)
    bins = [0, 6, 7, 8, 9, 10]
    labels = ['<6', '6-7', '7-8', '8-9', '>9']
    df['CGPA_Range'] = pd.cut(df['Current_CGPA'], bins=bins, labels=labels, include_lowest=True)

    # Clean CGPA_Change (from cell Ew_52s6X5om4)
    def clean_change(text):
        text = str(text).lower()
        text = unidecode.unidecode(text)
        text = re.sub(r'[^a-z\s]', '', text)
        text = text.strip()
        if 'mix' in text:
            return 'Mix'
        elif 'improved' in text:
            return 'Improved'
        elif 'declined' in text:
            return 'Declined'
        else:
            return 'Other'
    df['CGPA_Change'] = df['CGPA_Change'].apply(clean_change)

    # Ensure Current_CGPA is numeric and handle potential errors
    df['Current_CGPA'] = pd.to_numeric(df['Current_CGPA'], errors='coerce')

    # Filter out any rows that have missing values in these key columns before plotting
    df_filtered_violin = df.dropna(subset=['Current_CGPA', 'Academic_Satisfaction', 'Side_Interests']).copy()

    # Create the violin plot
    plt.figure(figsize=(12, 8)) # Adjust figure size for better readability
    sns.violinplot(x='Academic_Satisfaction', y='Current_CGPA', hue='Side_Interests', data=df_filtered_violin, palette='viridis', inner='quartile')
    plt.title('Current CGPA Distribution by Academic Satisfaction and Side Interests')
    plt.xlabel('Academic Satisfaction')
    plt.ylabel('Current CGPA')
    plt.legend(title='Side Interests', bbox_to_anchor=(1.05, 1), loc='upper left')
    plt.tight_layout() # Adjust layout to prevent labels from overlapping
    plt.show()
else:
    print("Violin plot not generated because the DataFrame is empty due to missing input file.")
```
